```python
from transformers import AutoTokenizer,AutoModelForSequenceClassification,pipeline
```

```python
tokenizer = AutoTokenizer.from_pretrained("mrm8488/distilroberta-finetuned-financial-news-sentiment-analysis")
```

```python
model = AutoModelForSequenceClassification.from_pretrained("mrm8488/distilroberta-finetuned-financial-news-sentiment-analysis")
```

```python
financial_Analysis = pipeline('sentiment-analysis',model=model,tokenizer=tokenizer)
```

```python
text = 'Stock price is soared by 20% by annual profit'
ans = financial_Analysis(text)
```

```python
ans
```

```python
text2 = "The stock dropped after the company announced weaker than expected revenue."
ans2 = financial_Analysis(text2)
ans2
```

```python
from transformers import AutoTokenizer,AutoModelForSequenceClassification,pipeline
```

```python
tokenizer2 = AutoTokenizer.from_pretrained("nlptown/bert-base-multilingual-uncased-sentiment")
model2 = AutoModelForSequenceClassification.from_pretrained("nlptown/bert-base-multilingual-uncased-sentiment")
sentiment_analysis = pipeline("sentiment-analysis",model=model2,tokenizer=tokenizer2)
text = [
    "This product is amazing, I really love it!",
    "The service was okay, nothing special.",
    "I am very disappointed, waste of money.",
    "यह प्रोडक्ट बहुत अच्छा है",  # Hindi (multilingual model)
    "Ce produit est terrible",    # French
    "El servicio fue excelente" ,  # Spanish
    "This is movie is waste of money "
]

for t in text:
    print(t,"->",sentiment_analysis(t))
```
